For Loop With Continue Statement

```python
for i in range(5):
  if(i==4):
    continue
  print("i=",i)
```

Nested for Loops

```python
for i in range(5):
  for i in range(i+1):
    print("*",end=" ")
  print()
```

Break Statement

```python
for i in range(3):
  for j in range(3):
    if(i!=j):
      print(i,j)
  print("======")
```
